## 提取 行业url到redis


```python
from spider.coroutine_crawl import construct_recruitment_url,RECRUITMENT_URL
from store.db.db_store import Mongo
m = Mongo()
find: list = m.find("industry_info", columns_conditions={"_id": 0, "third_industry": 1, "third_industry_url": 1})
for item in find:
    p = item.get("third_industry_url").split("-")[1].rstrip("/")
    construct_recruitment_url(RECRUITMENT_URL, position=p)
    item["third_industry_url"] = construct_recruitment_url(RECRUITMENT_URL, position=p)
    m.update_one("industry_info",{"third_industry":item.get("third_industry")},{"third_industry_url":item["third_industry_url"]})

find
```

```python
from store.db.db_store import get_redis

redis = get_redis()
```

```python
di = [(i["third_industry"],i["third_industry_url"]) for i in find]
di
```

```python
for i in di:
    redis.hset("industry_info",i[0],i[1])
```

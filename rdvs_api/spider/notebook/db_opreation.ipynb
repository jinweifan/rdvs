```python
jf = m.find('job_failed_log', columns_conditions={"_id": 0, "industry": 1})
len(jf), jf
```

```python
jfd = m.db['job_failed_log'].distinct('industry')
len(jfd), jfd
```

```python
## 更新failed表的url

def get_page_no(count: int):
    dvision = int(count / 30)
    mo = 1 if count % 30 else 0
    return dvision + mo


def update_failed_url(lis: list):
    for i in lis:
        inf: dict = m.db["job_failed_log"].find_one({"industry": i},
                                                    {"_id": 0, "industry": 1, "failed_url": 1, "crawled_count": 1})
        inf_url = inf.get("failed_url", None)
        inf_count = inf.get("crawled_count", None)
        sum = 0
        if "https://www.zhipin.com/web/geek/job" not in inf_url:
            url: dict = m.db["industry_info"].find_one({"third_industry": i}, {"_id": 0, "third_industry_url": 1})
            page_no = get_page_no(inf_count)
            updated_url = f'{url.get("third_industry_url")}&page={get_page_no(inf_count)}' if page_no else url.get(
                "third_industry_url")
            m.update_one('job_failed_log', {"industry": i}, {"failed_url": updated_url})
            # print(one)
            sum += 1
    print(f"更新了{sum}条数据")


update_failed_url(jfd)
```

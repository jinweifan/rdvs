```python
from store.db.db_store import Mongo

mongo = Mongo()
```

```python
info = mongo.find('industry_info', columns_conditions={'_id': 0})
info
```

```python
r_info = [
    {'industry': f'{i["first_industry"]}&{i["second_industry"]}&{i["third_industry"]}', 'url': i['third_industry_url']}
    for i in info]
r_info
```

```python
import ujson

redis.rpush('iis', *[ujson.dumps(i) for i in r_info])
```

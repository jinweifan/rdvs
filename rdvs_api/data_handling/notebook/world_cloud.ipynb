```python
import re

from store.db.db_store import Mongo

mongo = Mongo()
query: str = '[\s\S]*'
content = tuple(mongo.db['t2_job'].find({'third_industry': re.compile(fr'^{query}')}, {'_id': 0, 'skills': 1}))
content
```

```python
tp = [i.get('skills') for i in content]
tp
```

```python
string = ','.join(tp)
string
```

```python
string = string.replace(' ', '')
# string = string.replace('+', '\+')
string
```

```python
from  wordcloud import WordCloud
wc = WordCloud(font_path='/Users/reddyfan/code/proj/RDVS/rdvs/font/wryh.ttf', width=530, height=278,
               max_words=20,
               max_font_size=80, background_color='white').generate(string)
```

```python
import matplotlib.pyplot as plt

plt.imshow(wc, interpolation='bilinear')
plt.axis("off")
plt.show()
```

## 薪资转换函数

```python
def simple_salary_transform(salary: str):
    """ 简单薪资区间转换 如： 12-17K  """
    low, raw_high = salary.split('-')
    high = raw_high[:-1]
    return int(low), int(high)

def salary_range_handle(salary: str):
    """" 薪资转换函数 """
    if '薪' in salary:
        s1, r_xin = salary.split('·')
        xin = int(r_xin[:-1])
        r_low, r_high = simple_salary_transform(s1)
        low, high = round(r_low * xin / 12, 2), round(r_high * xin / 12, 2)
    elif '元' in salary:
        s1 = salary.split('/')[0]
        r_low, r_high = simple_salary_transform(s1)
        low, high = round(r_low * 20 / 1000, 2), round(r_high * 20 // 1000, 2)
    else:
        low, high = simple_salary_transform(salary)
    return round(low * 1000), round(high * 1000)
```

```python
raw = mongo.find('industry_info', columns_conditions={"_id": 0, "third_industry_url": 0})
df_industries = pd.DataFrame(raw)
df_industries.head()
```

```python
rg = [i[0] for i in df_industries.groupby(['first_industry', 'second_industry', 'third_industry'])]
```

```python
for i in rg[:5]:
    print(i)
```

### 将行业装换为字典

```python
def generate_industries(seq: list):
    dt = {}
    for i in seq:
        if dt.get(i[0], None):
            if dt[i[0]].get(i[1], None):
                dt[i[0]][i[1]] = dt[i[0]][i[1]] + [i[2]]
            else:
                dt[i[0]][i[1]] = [i[2]]
        else:
            dt[i[0]] = {i[1]: [i[2]]}
    return dt
```

```python
dt = generate_industries(rg)
dt
```

### 将行业字典处理为前台需要格式数据

```python
def generate_options(dt):
    if isinstance(dt, list):
        return [{'value': i, 'label': i} for i in dt]
    rt = []
    for k in dt:
        rt.append({'value': k, 'label': k, "children": generate_options(dt.get(k))})
    return rt
```

```python
industry_data = generate_options(dt)
industry_data
```

```python
redis.set('fontEnd:industry_data',ujson.dumps(industry_data))
```

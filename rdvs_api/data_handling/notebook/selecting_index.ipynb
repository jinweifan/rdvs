## 统计热门城市

### 加载数据

```python
import pandas as pd

from store.db.db_store import Mongo

mongo = Mongo()
raw_data = mongo.find('t2_job', columns_conditions={'job_area': 1, '_id': 0}, limit=5000)
df = pd.DataFrame(raw_data)
df
```

```python
df['city'] = df.job_area.map(lambda x: x.split('·')[0])
df.head(5)
```

```python
city_statistic = df.groupby('city').count()
city_statistic
```

```python
df.city.value_counts().to_dict()
```

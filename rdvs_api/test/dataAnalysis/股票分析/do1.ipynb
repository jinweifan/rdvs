```python
st_df = pd.read_csv('000004_st_daily_share.csv', parse_dates=['trade_date'])[['trade_date', 'close']]
agriculture_products_df = pd.read_csv('000061_农产品_daily_share.csv', parse_dates=['trade_date'])[
    ['trade_date', 'close']]
st_df = st_df.rename(columns={'close': 'software_close'})
agriculture_products_df = agriculture_products_df.rename(columns={'close': 'agriculture_products_close'})
data_merge = pd.merge(st_df, agriculture_products_df, how='inner', on='trade_date')
data_merge.to_excel('software_and_agriculture_products_share_price.xlsx')
data_merge
```

## 皮尔特系数——线性相关


```python
from scipy.stats import pearsonr
```

```python
corr = pearsonr(data_merge['software_close'], data_merge['agriculture_products_close'])
print('相关系数r值为' + str(corr[0]) + ', 显著水平P值为' + str(corr[1]))
```
